```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import cross_val_score
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score, f1_score
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor,RandomForestClassifier, GradientBoostingClassifier
import joblib
from sklearn.pipeline import Pipeline
```

```python
DATA_PATH = "cdr_naomie.xlsx"
TARGET_COL = "Avg throughput (ETSI A)"
SEUIL_DEBIT = 256000

print("Chargement du dataset...")
df = pd.read_excel(DATA_PATH)

print("\nDATASET BRUT")
print("Shape :", df.shape)
display(df.head())
```

```python
#SUPPRESSION DES COLONNES ENTIÈREMENT VIDES
print("\nSuppression des colonnes entièrement vides...")
df_clean = df.dropna(axis=1, how="all")
print("Colonnes supprimées :", set(df.columns) - set(df_clean.columns))
```

```python
# SUPPRESSION DES LIGNES ENTIÈREMENT VIDES
print("\nSuppression des lignes entièrement vides...")
df_clean = df_clean.dropna(axis=0, how="all")

print("\nDATASET APRÈS NETTOYAGE")
print("Shape :", df_clean.shape)
display(df_clean.head())
```

```python
#TRANSFORMATION DU DÉBIT EN SUCCÈS / ÉCHEC
print("\nTransformation du débit en classes Succès / Échec")

# Conversion en numérique (sécurité)
df_clean[TARGET_COL] = pd.to_numeric(df_clean[TARGET_COL], errors="coerce")

# Création de la variable de classification
df_clean["debit_class"] = df_clean[TARGET_COL].apply(
    lambda x: 1 if x >= SEUIL_DEBIT else 0
)
```

```python
#RÉSULTAT FINAL
print("\nDATASET FINAL AVEC CLASSIFICATION")
display(df_clean[[TARGET_COL, "debit_class"]].head(10))
```

```python
print("\n RÉPARTITION DES CLASSES")
print(df_clean["debit_class"].value_counts())

print("\n Analyse terminée avec succès")
```

```python
#SÉLECTION DES FEATURES ET CIBLES
FEATURES = ["AvgRSRP", "AvgRSRQ", "SINR"]
TARGET_REG = "Avg throughput (ETSI A)"
TARGET_CLF = "debit_class"
```

```python
# Vérification des colonnes
for col in FEATURES + [TARGET_REG, TARGET_CLF]:
    if col not in df_clean.columns:
        raise ValueError(f"Colonne manquante : {col}")
```

```python
# Suppression des lignes avec valeurs manquantes sur X ou y
df_model = df_clean[FEATURES + [TARGET_REG, TARGET_CLF]].dropna()
print("Dataset final pour la modélisation")
print("Shape :", df_model.shape)
display(df_model.head())
```

```python
# MATRICES X ET y
X = df_model[FEATURES]
y_reg = df_model[TARGET_REG]
y_clf = df_model[TARGET_CLF]

#SPLIT TRAIN / TEST
X_train, X_test, y_reg_train, y_reg_test = train_test_split(
    X, y_reg, test_size=0.2, random_state=42
)

_, _, y_clf_train, y_clf_test = train_test_split(
    X, y_clf, test_size=0.2, random_state=42
)
print("Train/Test split effectué")
```

```python
#NORMALISATION (POUR CERTAINS MODÈLES)
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Normalisation appliquée")
```

```python

print("\nDONNÉES PRÊTES")
print(f"Régression - Train : {X_train.shape}, Test : {X_test.shape}")
print(f"Classification - Train : {X_train.shape}, Test : {X_test.shape}")
```

```python
#MODELES DE REGRESSION 
print("RÉGRESSION — ENTRAÎNEMENT\n")

reg_pipelines = {
    "LinearRegression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LinearRegression())
    ]),
    "RandomForest": Pipeline([
        ("model", RandomForestRegressor(
            n_estimators=300,
            random_state=42
        ))
    ]),
    "GradientBoosting": Pipeline([
        ("model", GradientBoostingRegressor(
            random_state=42
        ))
    ])
}

reg_results = []
```

```python
for name, pipe in reg_pipelines.items():
    cv_rmse = -cross_val_score(
        pipe,
        X_train,
        y_reg_train,
        scoring="neg_root_mean_squared_error",
        cv=5
    ).mean()

    pipe.fit(X_train, y_reg_train)
    preds = pipe.predict(X_test)

    

    test_rmse = np.sqrt(mean_squared_error(y_reg_test, preds))

    r2 = r2_score(y_reg_test, preds)

    reg_results.append((name, cv_rmse, test_rmse, r2))

    print(f"{name}")
    print(f"  CV RMSE   : {cv_rmse:.2f}")
    print(f"  Test RMSE : {test_rmse:.2f}")
    print(f"  R²        : {r2:.3f}\n")
```

```python
reg_df = pd.DataFrame(
    reg_results,
    columns=["Modèle", "CV_RMSE", "Test_RMSE", "R2"]
).sort_values("CV_RMSE")

display(reg_df)

best_reg_name = reg_df.iloc[0]["Modèle"]
best_reg_model = reg_pipelines[best_reg_name]

joblib.dump(best_reg_model, "model/model_regression.joblib")

print(f"Meilleur modèle de régression : {best_reg_name}")
```

```python

#MODÈLES DE CLASSIFICATION
clf_models = {
    "LogisticRegression": LogisticRegression(max_iter=1000),
    "RandomForest": RandomForestClassifier(
        n_estimators=200, random_state=42
    ),
    "GradientBoosting": GradientBoostingClassifier(
        random_state=42
    )
}

clf_results = []
```

```python
print("ENTRAÎNEMENT – CLASSIFICATION\n")

for name, model in clf_models.items():
    # Cross-validation F1
    cv_f1 = cross_val_score(
        model,
        X_train,
        y_clf_train,
        scoring="f1",
        cv=5
    ).mean()

    # Entraînement final
    model.fit(X_train, y_clf_train)
    preds = model.predict(X_test)

    acc = accuracy_score(y_clf_test, preds)
    f1 = f1_score(y_clf_test, preds)

    clf_results.append((name, cv_f1, acc, f1))

    print(f"{name}")
    print(f"  CV F1 : {cv_f1:.3f}")
    print(f"  Accuracy : {acc:.3f}")
    print(f"  F1-score : {f1:.3f}\n")
```

```python
clf_df = pd.DataFrame(
    clf_results,
    columns=["Modèle", "CV_F1", "Accuracy", "F1"]
).sort_values("CV_F1", ascending=False)

display(clf_df)

# Meilleur modèle
best_clf_name = clf_df.iloc[0]["Modèle"]
best_clf_model = clf_models[best_clf_name]

# Sauvegarde
joblib.dump(best_clf_model, "model/model.joblib")

print(f"Meilleur modèle de classification : {best_clf_name}")
```
