```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.ensemble import GradientBoostingRegressor, GradientBoostingClassifier
from sklearn.ensemble import IsolationForest

from sklearn.metrics import (
    mean_squared_error,
    r2_score,
    accuracy_score,
    f1_score
)

import joblib

sns.set(style="whitegrid")
```

```python
DATA_PATH = "cdr_naomie.xlsx"

df = pd.read_excel(DATA_PATH)

print("Shape initiale :", df.shape)

# suppression colonnes vides
df = df.dropna(axis=1, how="all")

# suppression lignes vides
df = df.dropna(axis=0, how="all")

print("Shape après nettoyage :", df.shape)

#pd.set_option('display.max_rows', None)
#pd.set_option('display.max_columns', None)
#display(df)
df.head()
```

```python
TARGET = "Avg throughput (ETSI A)"

# conversion sécurité
df[TARGET] = pd.to_numeric(df[TARGET], errors="coerce")

# SEUIL RÉEL (2 Mbps -> bytes/s)
SEUIL_DEBIT = 256000

# classification
df["debit_class"] = (df[TARGET] >= SEUIL_DEBIT).astype(int)

# vérification
df[[TARGET, "debit_class"]].head()
```

```python
FEATURES = ["AvgRSRP", "AvgRSRQ", "SINR"]
TARGET_REG = "Avg throughput (ETSI A)"
TARGET_CLF = "debit_class"

df_model = df[FEATURES + [TARGET_REG, TARGET_CLF]].copy()


# conversion sécurité
df_model[TARGET_REG] = pd.to_numeric(df_model[TARGET_REG], errors="coerce")

# suppression des valeurs manquantes
df_model = df_model.dropna()

print("Dataset final :", df_model.shape)
df_model.describe()
```

```python
X = df_model[FEATURES]

y_reg = df_model[TARGET_REG]
y_clf = df_model[TARGET_CLF]
```

```python
X_train, X_test, y_reg_train, y_reg_test, y_clf_train, y_clf_test = train_test_split(
    X,
    y_reg,
    y_clf,
    test_size=0.2,
    random_state=42
)

print("Train shape :", X_train.shape)
print("Test shape :", X_test.shape)
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
reg_models = {
    "LinearRegression": LinearRegression(),

    "RandomForest": RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ),

    "GradientBoosting": GradientBoostingRegressor(
        random_state=42
    )
}
```

```python
reg_results = []

print("===== RÉGRESSION =====\n")

for name, model in reg_models.items():

    # entraînement
    model.fit(X_train, y_reg_train)

    # prédictions
    preds = model.predict(X_test)

    # métriques
    rmse = np.sqrt(mean_squared_error(y_reg_test, preds))
    r2 = r2_score(y_reg_test, preds)

    reg_results.append((name, rmse, r2))

    print(f"{name}")
    print(f"RMSE : {rmse:.2f}")
    print(f"R²   : {r2:.3f}\n")
```

```python
reg_df = pd.DataFrame(
    reg_results,
    columns=["Modèle", "RMSE", "R2"]
)

reg_df = reg_df.sort_values("R2", ascending=False)

reg_df
```

```python
clf_models = {
    "LogisticRegression": LogisticRegression(
        max_iter=1000
    ),

    "RandomForest": RandomForestClassifier(
        n_estimators=200,
        random_state=42
    ),

    "GradientBoosting": GradientBoostingClassifier(
        random_state=42
    )
}
```

```python
clf_results = []

print("===== CLASSIFICATION =====\n")

for name, model in clf_models.items():

    # entraînement
    model.fit(X_train, y_clf_train)

    # prédictions
    preds = model.predict(X_test)

    # métriques
    acc = accuracy_score(y_clf_test, preds)
    f1 = f1_score(y_clf_test, preds)

    clf_results.append((name, acc, f1))

    print(f"{name}")
    print(f"Accuracy : {acc:.3f}")
    print(f"F1-score : {f1:.3f}\n")
```

```python
clf_df = pd.DataFrame(
    clf_results,
    columns=["Modèle", "Accuracy", "F1"]
)

clf_df = clf_df.sort_values("F1", ascending=False)

clf_df
```

```python
best_reg_name = reg_df.iloc[0]["Modèle"]

print("Meilleur modèle de régression :", best_reg_name)
```

```python
best_clf_name = clf_df.iloc[0]["Modèle"]

print("Meilleur modèle de classification :", best_clf_name)
```
